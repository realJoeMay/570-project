```python
from ollama import chat

response = chat(
    model=llm_model,
    messages=[
        {
            "role": "user",
            "content": "Say hello and tell me what 2 + 2 is."
        }
    ],
)

print(response.message.content)
```

```python
from ollama import chat
from pydantic import BaseModel


class Ministry(BaseModel):
    name: str
    category: str


class Result(BaseModel):
    ministries: list[Ministry]


response = chat(
    model=llm_model,
    messages=[
        {
            "role": "user",
            "content": """
A Catholic parish has an Altar Server ministry and
a St. Vincent de Paul Society.

Extract the ministries and categorize them.
"""
        }
    ],
    format=Result.model_json_schema(),
    options={"temperature": 0},
)

result = Result.model_validate_json(response.message.content)

print(result)
```
